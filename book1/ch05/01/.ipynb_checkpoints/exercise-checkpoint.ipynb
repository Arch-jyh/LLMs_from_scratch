```python
import tiktoken
from previous_chapters import generate_text_simple

def text_to_token_ids(text,tokenizer):
    encoded = tokenizer.encode(text,allowed_special = {'<|endoftext|>'})
    #添加批维度
    encoded_tensor = torch.tensor(encoded).unsqueeze(0)
    return encoded_tensor

def token_ids_to_text(token_ids,tokenizer):
    #删除批维度
    flat = token_ids.squeeze(0)
    return tokenizer.decode(flat.tolist())

start_context = 'Every effort moves you'
tokenizer = tiktoken.get_encoding('gpt2')

token_ids = generate_text_simple(
    model = model,
    idx = text_to_token_ids(start_context,tokenizer),
    max_new_tokens = 10,
    context_size = GPT_CONFIG_124M['context_length']
)

out_text = token_ids_to_text(token_ids,tokenizer)

print('Out text:\n',out_text)
```

```python
import os
import urllib.request

file_path = "the-verdict.txt"
url = "https://raw.githubusercontent.com/rasbt/LLMs-from-scratch/main/ch02/01_main-chapter-code/the-verdict.txt"

if not os.path.exists(file_path):
    with urllib.request.urlopen(url) as rasponse:
        text_data = rasponse.read().decode('utf-8')
    with open(file_path,'w',encoding = 'utf-8') as file:
        file.write(text_data)
else:
    with open(file_path,'r',encoding = 'utf-8') as file:
        text_data = file.read()
```

```python
from previous_chapters import create_dataloader_v1

train_ratio = 0.90
split_idx = int(train_ratio * len(text_data))
train_data = text_data[:split_idx]
val_data = text_data[split_idx:]

torch.manual_seed(123)

train_loader = create_dataloader_v1(
    train_data,
    batch_size = 2,
    max_length = GPT_CONFIG_124M['context_length'],
    stride = GPT_CONFIG_124M['context_length'],
    #舍弃最后不满足一批的数据
    drop_last = True,
    #打乱顺序,每一个纪元读取,都会打乱顺序
    shuffle = True,
    #单进程加载数据
    num_workers = 0
)

val_loader = create_dataloader_v1(
    val_data,
    batch_size = 2,
    max_length = GPT_CONFIG_124M['context_length'],
    stride = GPT_CONFIG_124M['context_length'],
    #测试数据不舍弃
    drop_last = False,
    #测试数据不打乱
    shuffle = False,
    num_workers = 0
)
```

```python
#防止训练数据小到比context_length还小
if total_tokens * (train_ratio) < GPT_CONFIG_124M["context_length"]:
    print("Not enough tokens for the training loader. "
          "Try to lower the `GPT_CONFIG_124M['context_length']` or "
          "increase the `training_ratio`")
#防止验证数据小到比context_length还小
if total_tokens * (1-train_ratio) < GPT_CONFIG_124M["context_length"]:
    print("Not enough tokens for the validation loader. "
          "Try to lower the `GPT_CONFIG_124M['context_length']` or "
          "decrease the `training_ratio`")
```

```python
#calc是calculate的意思
def calc_loss_batch(input_batch,target_batch,model,device):
    input_batch,target_batch = input_batch.to(device),target_batch.to(device)
    #logits是原始的分
    logits = model(input_batch)
    #cross_entropy里面自带softmax
    loss = torch.nn.functional.cross_entropy(logits.flatten(0,1),target_batch.flatten())
    return loss


def calc_loss_loader(data_loader,model,device,num_batches = None):
    #初始化累计损失,用浮点数初始化
    total_loss = 0.
    #返回数据加载器的批次总数
    if len(data_loader) == 0:
        #获取nan,nan归于浮点数,所以用float获取(float规范的特殊格式)
        return float('nan')
    #如果没有指定遍历多少个batch,那么遍历所有的batch
    elif num_batches is None:
        num_batches = len(data_loader)
    else:
        #防止num_batches数超过data_loader
        num_batches = min(num_batches,len(data_loader))
    for i,(input_batch,target_batch) in enumerate(data_loader):
        if i < num_batches:
            loss = calc_loss_batch(input_batch,target_batch,model,device)
            total_loss += loss.item()
        else:
            break

    return total_loss / num_batches
```

```python
device = torch.device("mps" if torch.backends.mps.is_available() else "cpu")
#数据和model不在一起会报错
model.to(device)

torch.manual_seed(123)

with torch.no_grad():
    train_loss = calc_loss_loader(train_loader,model,device)
    val_loss = calc_loss_loader(val_loader,model,device)

print('Training loss:',train_loss)
print('Validation loss:',val_loss)
```

```python
#eval_freq:评估频率,多少step评估一次损失
#eval_iter:评估采样批次数量,每一次评估取多少批次
#start_context:起始文本,模型生成文本,供人检查
#tokenizer:分词器
def train_model_simple(model,train_loader,val_loader,optimizer,device,num_epochs,
                       eval_freq,eval_iter,start_context,tokenizer):

    #训练损失记录列表,验证损失记录列表,累计见过的 Token 数追踪列表
    train_losses,val_losses,track_tokens_seen = [],[],[]
    #初始化全局记数
    #累计模型见过的 Token 总量,全局训练步数从 -1 开始
        #进入循环后,step = 0,从0开始触发初始评估
    tokens_seen,global_step = 0,-1

    for epoch in range(num_epochs):
        #训练模式
        model.train()

        for input_batch,target_batch in train_loader:
            #重置梯度
            optimizer.zero_grad()
            loss = calc_loss_batch(input_batch,target_batch,model,device)
            loss.backward()
            #参数更新,用step命名意思是沿着梯度迈出一步(step)
            optimizer.step()
            #累计消耗的token
            tokens_seen += input_batch.numel()
            global_step += 1

            #评估频率判断
            if global_step % eval_freq == 0:
                #评估模型
                train_loss,val_loss = evaluate_model(
                    model,train_loader,val_loader,device,eval_iter)
                train_losses.append(train_loss)
                val_losses.append(val_loss)
                #记录这次评估参与了多少的token
                track_tokens_seen.append(tokens_seen)
                #d表示Decimal,十进制整数 0表示补零 6表示6位
                print(f'Ep {epoch+1} (Step {global_step:06d}):'
                      f'Train loss {train_loss:.3f},Val loss {val_loss:.3f}')
        
        #给一个例子并打印结果
        generate_and_print_sample(
            model,tokenizer,device,start_context
        )
    return train_losses, val_losses, track_tokens_seen


def evaluate_model(model,train_loader,val_loader,device,eval_iter):
    #评估模式 ,禁用dropout
    model.eval()
    with torch.no_grad():
        train_loss = calc_loss_loader(train_loader,model,device,num_batches = eval_iter)
        val_loss = calc_loss_loader(val_loader,model,device,num_batches = eval_iter)
    #恢复训练模式
    model.train()
    return train_loss,val_loss

def generate_and_print_sample(model,tokenizer,device,start_context):
    model.eval()
    #位置嵌入的参数的第0维度,对应了输入的序列上限
    context_size = model.pos_emb.weight.shape[0]
    encoded = text_to_token_ids(start_context,tokenizer).to(device)
    with torch.no_grad():
        #context_size限制了最大输入词元id
        #token_ids包含了输入的encoded和输出的50词
        token_ids = generate_text_simple(
            model = model,
            idx = encoded,
            max_new_tokens = 50,
            context_size = context_size
        )
        decoded_text = token_ids_to_text(token_ids,tokenizer)
        #换行变为空格,紧凑打印
        print(decoded_text.replace('\n',' '))
    model.train()
```

```python
torch.manual_seed(123)
model = GPTModel(GPT_CONFIG_124M)
model.to(device)
optimizer = torch.optim.AdamW(model.parameters(),lr = 0.0004,weight_decay = 0.1)

num_epochs = 10
train_losses,val_losses,tokens_seen = train_model_simple(
    model,train_loader,val_loader,optimizer,device,
    num_epochs = num_epochs,eval_freq = 5,eval_iter = 5,
    start_context = 'Every effort moves you',tokenizer = tokenizer
)
```

```python
import matplotlib.pyplot as plt
#控制最大刻度的定位器
from matplotlib.ticker import MaxNLocator

def plot_losses(epochs_seen,tokens_seen,train_losses,val_losses):
    fig,ax1 = plt.subplots(figsize = (5,3))

    ax1.plot(epochs_seen,train_losses,label = 'Training loss')
    ax1.plot(epochs_seen,val_losses,linestyle = '-.',label = 'Validation loss')
    ax1.set_xlabel('Epochs')
    ax1.set_ylabel('Loss')
    ax1.legend(loc = 'upper right')
    #MaxNLocator控制刻度为合理在合理范围里,integer控制是不是整数
        #MaxNLocator用在传输数据之后,这时候有最大和最小值,可以控制到整数了
    ax1.xaxis.set_major_locator(MaxNLocator(integer = True))
    #共享y的孪生坐标系,在顶部生成x轴
    ax2 = ax1.twiny()
    #重复画线了,不过alpha为0,全透明了,为了绘制tokens_seen
    ax2.plot(tokens_seen,train_losses,alpha = 0)
    ax2.set_xlabel('Tokens seen')

    fig.tight_layout()
    plt.savefig('loss_plot.pdf')
    plt.show()

#实际上这个错了一位,演示可忽略
epochs_tensor = torch.linspace(0,num_epochs,len(train_losses))
plot_losses(epochs_tensor,tokens_seen,train_losses,val_losses)
```

```python
def print_sampled_tokens(probas):
    torch.manual_seed(123)
    #python自动忽略数字中的'_','_'方便分割
    sample = [torch.multinomial(probas,num_samples = 1).item() for i in range(1_000)]
    #bincount用于统计,吧tensor的元素自动统计出'直方图'的数据
        #用id作为索引,数值为频数
    sampled_ids = torch.bincount(torch.tensor(sample))
    for i,freq in enumerate(sampled_ids):
        #如果记数没有最后的索引,那么这里也不会显示最后索引的记数
        print(f'{freq} x {inverse_vocab[i]}')

print_sampled_tokens(probas)
```

```python
def softmax_with_temperature(logits,temperature):
    scaled_logits = logits / temperature
    return torch.softmax(scaled_logits,dim = 0)

temperatures = [1,0.1,5]

scaled_probas = [softmax_with_temperature(next_token_logits,T) for T in temperatures]
scaled_probas
```

```python
x = torch.arange(len(vocab))
bar_width = 0.15

fig,ax = plt.subplots(figsize = (5,3))
for i,T in enumerate(temperatures):
    rects = ax.bar(x + i * bar_width,scaled_probas[i],bar_width,label = f'Temperature = {T}')

ax.set_ylabel('Probability')
ax.set_xticks(x)
ax.set_xticklabels(vocab.keys(),rotation = 90)
ax.legend()

plt.tight_layout()
plt.savefig('temperature-plot.pdf')
plt.show()
```

```python
#eos_id是文本结束符号,不设置将会完整的跑完整个预配置的token数量
#idx是整个输入
def generate(model,idx,max_new_tokens,context_size,temperature = 0.0,top_k = None,eos_id = None):
    for _ in range(max_new_tokens):
        #只关注输入的最后的context_size长度后的数据作为输入,截取可以接收的长度
        idx_cond = idx[:,-context_size:]
        with torch.no_grad():
            logits = model(idx_cond)
        #提取出最后一个预测
        logits = logits[:,-1,:]

        #没有else如果判断为False,也会继续执行
        if top_k is not None:
            #返回的第一个是概率,第二个是position索引
            top_logits,_ = torch.topk(logits,top_k)
            min_val = top_logits[:,-1]
            logits = torch.where(logits < min_val,torch.tensor(float('-inf')).to(logits.device),logits)

        if temperature > 0.0:
            logits = logits / temperature

            probs = torch.softmax(logits,dim = -1)

            #multinomial提出出来的是位置索引
            idx_next = torch.multinomial(probs,num_samples = 1)

        else:
            idx_next = torch.argmax(logits,dim = -1,keepdim = True)

        #触发了结束词,结束循环
        if idx_next == eos_id:
            break
        idx = torch.cat((idx,idx_next),dim = 1)

    return idx
```

```python
def assign(left,right):
    #tensorflow返回的可能是numpy数组
    #numpy的.shape返回元组,torch的.shape返回tensor.size
        #tensor.size继承的元祖,元祖和元祖,元祖和子类是可以比较的
    if left.shape != right.shape:
        raise ValueError(f'Shape mismatch. Left:{left.shape},Right:{right.shape}')
    #转换为正式的模型权重
    return torch.nn.Parameter(torch.tensor(right))
```

```python
import numpy as np
```

```python
def load_weights_into_gpt(gpt,params):
    gpt.pos_emb.weight = assign(gpt.pos_emb.weight,params['wpe'])
    gpt.tok_emb.weight = assign(gpt.tok_emb.weight,params['wte'])

    #遍历每一个transformer模块
    for b in range(len(params['blocks'])):
        #把拼接好的qkv分开
        q_w,k_w,v_w = np.split(
            #c_attn的c是Combined,三合一的注意力层  w是权重
            (params['blocks'][b]['attn']['c_attn'])['w'],3,axis = -1)
        gpt.trf_blocks[b].att.W_query.weight = assign(
            gpt.trf_blocks[b].att.W_query.weight,q_w.T)
        gpt.trf_blocks[b].att.W_key.weight = assign(
            gpt.trf_blocks[b].att.W_key.weight,k_w.T)
        gpt.trf_blocks[b].att.W_value.weight = assign(
            gpt.trf_blocks[b].att.W_value.weight,v_w.T)

        q_b,k_b,v_b = np.split(
            (params['blocks'][b]['attn']['c_attn'])['b'],3,axis = -1)
        gpt.trf_blocks[b].att.W_query.bias = assign(
            gpt.trf_blocks[b].att.W_query.bias,q_b.T)
        gpt.trf_blocks[b].att.W_key.bias = assign(
            gpt.trf_blocks[b].att.W_key.bias,k_b.T)
        gpt.trf_blocks[b].att.W_value.bias = assign(
            gpt.trf_blocks[b].att.W_value.bias,v_b.T)

        #.T是tensorFlow和PyTorch的权重约定相反
        #投影层,把qkv拼接
        gpt.trf_blocks[b].att.out_proj.weight = assign(
            gpt.trf_blocks[b].att.out_proj.weight,
            params['blocks'][b]['attn']['c_proj']['w'].T)
        gpt.trf_blocks[b].att.out_proj.bias = assign(
            gpt.trf_blocks[b].att.out_proj.bias,
            params['blocks'][b]['attn']['c_proj']['b'].T)

        #前馈网络层
        #layers[1]是GELU(),只有公式,所以不需要加载覆盖
        gpt.trf_blocks[b].ff.layers[0].weight = assign(
            gpt.trf_blocks[b].ff.layers[0].weight, 
            params["blocks"][b]["mlp"]["c_fc"]["w"].T)
        gpt.trf_blocks[b].ff.layers[0].bias = assign(
            gpt.trf_blocks[b].ff.layers[0].bias, 
            params["blocks"][b]["mlp"]["c_fc"]["b"])
        gpt.trf_blocks[b].ff.layers[2].weight = assign(
            gpt.trf_blocks[b].ff.layers[2].weight, 
            params["blocks"][b]["mlp"]["c_proj"]["w"].T)
        gpt.trf_blocks[b].ff.layers[2].bias = assign(
            gpt.trf_blocks[b].ff.layers[2].bias, 
            params["blocks"][b]["mlp"]["c_proj"]["b"])

        #归一化层,这是前置的
        #一个是注意力的一个是前馈神经网络的
        gpt.trf_blocks[b].norm1.scale = assign(
            gpt.trf_blocks[b].norm1.scale, 
            params["blocks"][b]["ln_1"]["g"])
        gpt.trf_blocks[b].norm1.shift = assign(
            gpt.trf_blocks[b].norm1.shift, 
            params["blocks"][b]["ln_1"]["b"])
        gpt.trf_blocks[b].norm2.scale = assign(
            gpt.trf_blocks[b].norm2.scale, 
            params["blocks"][b]["ln_2"]["g"])
        gpt.trf_blocks[b].norm2.shift = assign(
            gpt.trf_blocks[b].norm2.shift, 
            params["blocks"][b]["ln_2"]["b"])

    #最终归一化
    gpt.final_norm.scale = assign(gpt.final_norm.scale, params["g"])
    gpt.final_norm.shift = assign(gpt.final_norm.shift, params["b"])
    #线性输出
    gpt.out_head.weight = assign(gpt.out_head.weight, params["wte"])

load_weights_into_gpt(gpt, params)
gpt.to(device);
```
