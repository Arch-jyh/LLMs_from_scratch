```python
import time
import torch

# 自动检测加速设备（在你的 Mac 上会使用 'mps'）
device = torch.device(
    "mps"
    if torch.backends.mps.is_available()
    else "cuda"
    if torch.cuda.is_available()
    else "cpu"
)
print("当前测试的设备是:", device)
```

```python
N = 10000

# 1. 创建 CPU 上的矩阵并测试时间
a_cpu = torch.randn(N, N)
b_cpu = torch.randn(N, N)
print("--- CPU 矩阵乘法时间 ---")
%timeit a_cpu @ b_cpu

# 2. 创建 GPU(MPS) 上的矩阵并测试时间
a_gpu = a_cpu.to(device)
b_gpu = b_cpu.to(device)
print(f"--- GPU({device}) 矩阵乘法时间 ---")
%timeit a_gpu @ b_gpu
```
